# 01 - Khám phá dữ liệu (EDA)

Mục tiêu: hiểu dữ liệu trước khi train. Trả lời các câu hỏi:
1. Tỷ lệ sinh viên học yếu là bao nhiêu? Dữ liệu có mất cân bằng không?
2. Có giá trị thiếu (missing) không?
3. Đặc trưng nào khác biệt rõ nhất giữa nhóm học yếu và không học yếu?

Chạy trước: `python ml/scripts/generate_synthetic_data.py`

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/synthetic/students_synthetic.csv")
df.head()
```

```python
print("Số dòng:", len(df))
print("Giá trị thiếu theo cột:")
print(df.isna().sum())
df.describe().round(2)
```

```python
# 1. Mất cân bằng lớp
df["at_risk"].value_counts().plot.bar(title="Số sinh viên theo nhãn (1 = học yếu)")
plt.show()
print(f"Tỷ lệ học yếu: {df['at_risk'].mean():.1%}")
```

```python
# 3. So sánh trung bình từng đặc trưng giữa 2 nhóm
features = [
    "gpa_prev", "attendance_rate", "midterm_avg", "failed_courses_prev",
    "credits_registered", "assignment_submission_rate", "lms_logins_per_week",
]
df.groupby("at_risk")[features].mean().T.round(2)
```

```python
fig, axes = plt.subplots(2, 4, figsize=(16, 7))
for ax, col in zip(axes.flat, features):
    df.boxplot(column=col, by="at_risk", ax=ax)
    ax.set_title(col)
    ax.set_xlabel("at_risk")
axes.flat[-1].axis("off")
plt.suptitle("Phân bố đặc trưng theo nhãn")
plt.tight_layout()
plt.show()
```

```python
# Tương quan giữa các đặc trưng và nhãn
df[features + ["at_risk"]].corr()["at_risk"].drop("at_risk").sort_values()
```
